```python
def UnitUsed():
    while True:
        previous =  (input("Enter the previous meter reading: "))
        current =  (input("Enter the current meter reading: "))

        if previous.isdigit() and current.isdigit():
            previous = int(previous)
            current = int(current)

            if current > previous:
                return current - previous
            else:
                print("Enter the current amount geater than previous amount.")

        else:
            print("Enter the current amount and previous amount in whole numbers.")
            
```

```python
def Slab_Charges(unit):
    charge1 = 0
    charge2 = 0
    charge3 = 0
    charge4 = 0
    charge5 = 0

    if unit <=100:
        charge1 = unit*16
    else:
        charge1 = 100 *16

    if unit > 100:
        if unit <= 200:
            charge2 = (unit - 100) * 22
        else:
            charge2 = 100 * 22

    if unit > 200:
        if unit <= 300:
            charge3 = (unit - 200)* 27
        else:
            charge3 = 100 * 27

    if unit > 300:
        if unit <= 400:
            charge4 = (unit - 300) * 32
        else:
            charge4 = 100 * 32

    if unit > 400:
        charge5 = (unit - 400) * 38

    print("\n       slab price")
    print(f"1-100 units: Rs.{charge1:,.2f}")
    print(f"101-200 units: Rs.{charge2:,.2f}")
    print(f"201-300 units: Rs.{charge3:,.2f}")
    print(f"301-400 units: Rs.{charge4:,.2f}")
    print(f"Above 400 units: Rs.{charge5:,.2f}")

    return charge1 + charge2 + charge3 + charge4 + charge5
    
```

```python
def total_Bill(energy):
    FixCharges = 150
    tax = (FixCharges + energy) * 0.18
    late = energy * 0.10

    BeforeDue = energy + tax + FixCharges
    AfterDue = BeforeDue + late

    print("\n         Bill Detail ")
    print(f"Energy charges : Rs.{energy:,.2f}")
    print(f"Tax : Rs.{tax:,.2f}")
    print(f"Fix charges: Rs.{FixCharges:,.2f}")
    print(f"Bill Before Due Date: Rs.{BeforeDue:,.2f}")
    print(f"Late Charges:Rs.{late:,.2f}")
    print(f"Bill After Due Date: {AfterDue:,.2f}")

    return BeforeDue
```

```python
def PrintBill(unit,name):
    print("\n           ELECTRICITY BILL") 
    print(f"Customer Name: {name}")
    print(f"Unit Used: {unit}")

    energy = Slab_Charges(unit)
    bill = total_Bill(energy)


    if unit <= 100:
        print(f"\nTip: {101-unit} units left to reach the next slab.")
    elif unit <= 200:
        print(f"\nTip: {201-unit} units left to reach the next slab.")
    elif unit<=300:
        print(f"\nTip: {301-unit} units left to reach the next slab")
    elif unit <= 400:
        print(f"\nTip: {401-unit} units left to reach the next slab")
    else:
        print("You are Already in next Highest slab.")

    return bill
```

```python
count = 0
total_unit = 0
highest_bill = 0
highest_customer = ""
while True:
    name = input("Enter Customer name: ")

    unit = UnitUsed()
    bill = PrintBill(unit,name)

    count += 1 
    total_unit += unit

    if bill > highest_bill:
        highest_bill = bill
        highest_customer = name

    again = ("Do you want to calculate another bill.(yes/no): ")
    if again.lower != "yes":
        break

    
```
