Write grade(marks) that returns A, B, C, D or F

```python
def grade(marks):
    """find the grade by checking the marks"""
    if marks < 0 or marks > 100:
        return "Invalid marks"
    elif marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    elif marks >= 70:
        return "C"
    elif marks >= 60:
        return "D"
    else:
        return "F"


marks = float(input("Enter your marks: "))
print("your grade is: ",grade(marks))
```

Write a recursive function factorial(n). 

```python
def factorial(number):
    """find the factorial of the number by recursion."""
    if number < 0:
        return "Please enter the positive number."

    if number == 0 or number == 1:
        return 1
    
    return number*factorial(number-1)
num = int(input("Enter the positive number: "))
print("factorial of number = ",factorial(num))
```

Write a recursive function sum_of_digits(n) (for example 472 → 13). 

```python
def sum_of_digits(number): 
    """find the sum of the digits of the number by recursion"""
    if number == 0:
        return 0
    return (number%10)+ sum_of_digits(number//10)

print("sum of the number is = ",sum_of_digits(472))
```

Write is_prime(n) and use it to print all prime numbers between 1 and 100. 

```python
def is_prime(n):
    if n < 2:
        return False
    i = 2
    while i*i <= n:
        if n % i == 0:
            return False
        i += 1
    return True

for n in range(2,101):
    if is_prime(n):
        print(n,end=" ")
```
