Take n from the user and print the sum and the average of the numbers from 1 to n.

```python
num = int (input("Enter the number: "))
print("input is = ",num)
sum = 0
for i in range(1,num+1):
    sum = sum + i
print(f"Sum = {sum}")
print(f"Average = {sum/num}")    
```

Count the digits of a number using a while loop (for example 40531 → 5 digits).

```python
num = int(input("Enter the positive number: "))
print("input = ",num)
count = 0
if num == 0:
    print("There are only 1 digit in the number")

elif num < 0:
    print("Please enter the positive number.")

else:
    while num > 0:
        num = num // 10
        count += 1

    print(f"There are {count} digit in the number. ")
    
```

Take n from the user and print this pattern with nested loops

```python
rows = int (input("Enter the number of rows: "))
print("input =",rows)
for i in range(1, rows + 1):
    for j in range(1,i+1):
        print("*",end=" ")
        
    print()   
    
```

Guess-the-number game: keep asking until the user guesses the secret number, use break,
and print the number of attempts

```python
import random
random_number = random.randint(1,10)
attempt = 0
while True:
    guess = int (input("Guess number the number: "))
    attempt += 1
    if guess == random_number:
        print(f"You guess the number in {attempt} attempts.")
        print("Your guess is =", guess)
        break
    elif guess<random_number:
        print("Your guess is samller then the actual number. Try somthing Big.")
    else:
        print("Your guess is bigger then the actual number. Try somthing Small.")  
    print("Your guess is:", guess)      
```

Count the vowels in a sentence entered by the user

```python
sentence = input("Enter the sentence: ")
print("input = ", sentence)
vowel = "aeiouAEIOU"
count = 0
for char in sentence:
    if char in vowel:
        count += 1

print(f"There are {count} vowel in this sentence.")
```

Reverse a word and check whether it is a palindrome (for example "madam").

```python
word = input("Enter the word to check weather it is palindrome or not: ")
print("Input = ",word)
reverse = ""
for char in word:
    reverse = char+reverse     # Another way to reverse is reverse = word[::-1]
if reverse == word:
    print("The word is palindrome.")
else:
    print("The word is not pallindrome.")        
```

Count the words in a sentence.

```python
sentence = input("Write the sentence: ")
print("input =", sentence )
count = 0
if len(sentence) == 0:
    print("Please wirte the sentence never let it empty.")
else:
    for char in sentence:
        if char == " ":
            continue
        else:
            count += 1    

print(f"There are {count} letter in this sentence.")
```
