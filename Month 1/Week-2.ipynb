Mini Calculator

```python
num1 = int (input("Enter the first number: "))
operator = input("Enter the operator (+, -, x, /):")
num2 = int(input("Enter the second number: "))
if(operator == "+"):
    print(f"{num1} + {num2} = {num1 + num2}")
elif(operator == "-"):
    print(f"{num1} - {num2} = {num1 - num2}") 
elif(operator == "X" or operator == "x"):
    print(f"{num1} x {num2} = {num1 * num2}")
elif(operator == "/"):
    if num2 == 0:
        print("Sorry dividing by 0 is not possible 😔 Please Try Again.")
    else:
        print(f"{num1} ÷ {num2} = {num1 / num2}")
else:
    print("Invalid operator")                   
```

Finding the leap year

```python
year = int (input("Enter the Year: "))
if(year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(f"{year} is a leap year.")
else:
    print(f"{year} is not a leap year.")    
```

Shop Discount Boundary Test

```python
bill = int(input("Enter your bill: "))
if(bill > 0):
    if bill > 5000:
        bill = bill-(bill/100)*10     # customer got 10% discount
        print(f"You have to pay only {bill} Rs.")
    else:
        print(f"You have to pay only {bill} Rs.")
else:
    print("Invalid bill.\nPlease Try Again.")        
```
