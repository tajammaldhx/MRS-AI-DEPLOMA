Celsius to Fahrenheit. Take a temperature in °C from the user and print it in °F.
Formula: F = C × 9 / 5 + 32.

```python
temperature = int(input("Enter the Temperature in celsius: "))
fahrenheit = (temperature * 9/5) + 32
print (f"Temperature in Fahrenheit: {fahrenheit}")
```

Split the bill. Take the total bill and the number of friends. Print each person’s share in whole rupees ( // ) and the rupees left over (%).

```python
bill = int (input("Enter the amount : "))
frnd = int (input("Enter the number of friends: "))
PerPerson = bill//frnd
reminder = bill%frnd
print(f"Bill per person have to pay : {PerPerson} Rs")
print(f"Remaining Amount : {reminder} Rs")
```
